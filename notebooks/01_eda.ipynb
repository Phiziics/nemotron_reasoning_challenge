```python
from pathlib import Path
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
```

```python
train_path = Path("../data/raw/train.csv")

print("File exists:", train_path.exists())
print("Path:", train_path)

df = pd.read_csv(train_path)

print("Shape:", df.shape)
display(df.head())
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes)
```

```python
missing = (
    df.isna()
    .sum()
    .reset_index()
    .rename(columns={"index": "column", 0: "missing_count"})
)

missing["missing_percent"] = (missing["missing_count"] / len(df) * 100).round(2)

display(missing.sort_values("missing_count", ascending=False))
```

```python
duplicate_rows = df.duplicated().sum()

print("Duplicate rows:", duplicate_rows)
print("Duplicate percent:", round(duplicate_rows / len(df) * 100, 2))
```
