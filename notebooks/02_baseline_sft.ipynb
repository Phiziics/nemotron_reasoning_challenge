```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
```

```python
PROJECT_DIR = Path("../")

RAW_DATA_DIR = PROJECT_DIR / "data" / "raw"
PROCESSED_DATA_DIR = PROJECT_DIR / "data" / "processed"

TRAIN_PATH = RAW_DATA_DIR / "train.csv"
TEST_PATH = RAW_DATA_DIR / "test.csv"

PROCESSED_DATA_DIR.mkdir(parents=True, exist_ok=True)

print("Project exists:", PROJECT_DIR.exists())
print("Train exists:", TRAIN_PATH.exists())
print("Test exists:", TEST_PATH.exists())
```

```python
train_df = pd.read_csv(TRAIN_PATH)

print("Train shape:", train_df.shape)
display(train_df.head())
print("Columns:", train_df.columns.tolist())
```

```python
print("Rows:", train_df.shape[0])
print("Columns:", train_df.shape[1])

print("\nData types:")
display(train_df.dtypes)

missing = train_df.isna().sum().reset_index()
missing.columns = ["column", "missing_count"]
missing["missing_percent"] = (missing["missing_count"] / len(train_df) * 100).round(2)

display(missing.sort_values("missing_count", ascending=False))

print("Duplicate rows:", train_df.duplicated().sum())
```

```python
def find_column(df, possible_names):
    columns_lower = {col.lower(): col for col in df.columns}
    
    for name in possible_names:
        if name.lower() in columns_lower:
            return columns_lower[name.lower()]
    
    return None


prompt_col = find_column(
    train_df,
    ["prompt", "question", "puzzle", "input", "problem"]
)

answer_col = find_column(
    train_df,
    ["answer", "target", "output", "solution", "label"]
)

print("Prompt column:", prompt_col)
print("Answer column:", answer_col)

if prompt_col is None:
    raise ValueError(f"Could not find prompt column. Columns: {train_df.columns.tolist()}")

if answer_col is None:
    raise ValueError(f"Could not find answer column. Columns: {train_df.columns.tolist()}")
```

```python
pd.set_option("display.max_colwidth", 1000)

display(train_df[[prompt_col, answer_col]].head(10))
```

```python
baseline_df = train_df[[prompt_col, answer_col]].copy()

baseline_df.columns = ["prompt", "answer"]

baseline_df["prompt"] = baseline_df["prompt"].astype(str).str.strip()
baseline_df["answer"] = baseline_df["answer"].astype(str).str.strip()

baseline_df = baseline_df.dropna()
baseline_df = baseline_df.drop_duplicates()

print("Clean baseline shape:", baseline_df.shape)
display(baseline_df.head())
```

```python
SYSTEM_PROMPT = """You are solving a reasoning puzzle.
Return only the final answer."""


def format_sft_text(prompt, answer):
    return f"""{SYSTEM_PROMPT}

Puzzle:
{prompt}

Answer:
{answer}"""


baseline_df["text"] = baseline_df.apply(
    lambda row: format_sft_text(row["prompt"], row["answer"]),
    axis=1
)

display(baseline_df[["text"]].head())
```

```python
baseline_df["prompt_length"] = baseline_df["prompt"].str.len()
baseline_df["answer_length"] = baseline_df["answer"].str.len()
baseline_df["text_length"] = baseline_df["text"].str.len()

display(
    baseline_df[
        ["prompt_length", "answer_length", "text_length"]
    ].describe()
)
```

```python
baseline_df["prompt_length"] = baseline_df["prompt"].str.len()
baseline_df["answer_length"] = baseline_df["answer"].str.len()
baseline_df["text_length"] = baseline_df["text"].str.len()

display(
    baseline_df[
        ["prompt_length", "answer_length", "text_length"]
    ].describe()
)
```

```python
plt.figure(figsize=(8, 5))
plt.hist(baseline_df["text_length"], bins=40)
plt.title("SFT Text Length Distribution")
plt.xlabel("Text Length")
plt.ylabel("Count")
plt.show()
```

```python
train_sft_df, val_sft_df = train_test_split(
    baseline_df,
    test_size=0.05,
    random_state=42,
    shuffle=True
)

print("Train SFT shape:", train_sft_df.shape)
print("Validation SFT shape:", val_sft_df.shape)
```

```python
train_sft_path = PROCESSED_DATA_DIR / "train_sft.csv"
val_sft_path = PROCESSED_DATA_DIR / "val_sft.csv"
full_sft_path = PROCESSED_DATA_DIR / "full_sft.csv"

train_sft_df.to_csv(train_sft_path, index=False)
val_sft_df.to_csv(val_sft_path, index=False)
baseline_df.to_csv(full_sft_path, index=False)

print("Saved:")
print(train_sft_path)
print(val_sft_path)
print(full_sft_path)
```

```python
train_jsonl_path = PROCESSED_DATA_DIR / "train_sft.jsonl"
val_jsonl_path = PROCESSED_DATA_DIR / "val_sft.jsonl"
full_jsonl_path = PROCESSED_DATA_DIR / "full_sft.jsonl"

train_sft_df[["text"]].to_json(
    train_jsonl_path,
    orient="records",
    lines=True,
    force_ascii=False
)

val_sft_df[["text"]].to_json(
    val_jsonl_path,
    orient="records",
    lines=True,
    force_ascii=False
)

baseline_df[["text"]].to_json(
    full_jsonl_path,
    orient="records",
    lines=True,
    force_ascii=False
)

print("Saved JSONL:")
print(train_jsonl_path)
print(val_jsonl_path)
print(full_jsonl_path)
```
